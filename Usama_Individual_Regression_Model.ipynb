## Why I chose this model

I used regression because our group EDA showed that demand is strongly affected by promotions and pricing. Unlike ARIMA or Exponential Smoothing, regression can directly model how external features influence weekly food orders.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import mean_absolute_error, mean_squared_error

sns.set_style('whitegrid')
print('Libraries imported successfully')
```

```python
train_df = pd.read_csv('Datasets/train.csv')
meal_info_df = pd.read_csv('Datasets/meal_info.csv')
center_info_df = pd.read_csv('Datasets/fulfilment_center_info.csv')

df = train_df.merge(meal_info_df, on='meal_id', how='left')
df = df.merge(center_info_df, on='center_id', how='left')
df = df.sort_values(['center_id', 'meal_id', 'week']).copy()

print('Merged shape:', df.shape)
df.head()
```

```python
df['lagged_orders_1week'] = df.groupby(['center_id', 'meal_id'])['num_orders'].shift(1)
df['rolling_avg_7week'] = df.groupby(['center_id', 'meal_id'])['num_orders'].transform(lambda x: x.rolling(window=7, min_periods=1).mean())
df['lagged_orders_1week'] = df['lagged_orders_1week'].fillna(df['num_orders'].mean())
df['rolling_avg_7week'] = df['rolling_avg_7week'].fillna(df['num_orders'].mean())

meal_dummies = pd.get_dummies(df['category'], prefix='meal_cat')
center_dummies = pd.get_dummies(df['center_type'], prefix='center_type')
df = pd.concat([df, meal_dummies, center_dummies], axis=1)

print('Feature engineering complete')
```

```python
split_week = 130
train_data = df[df['week'] <= split_week].copy()
test_data = df[df['week'] > split_week].copy()

exclude_cols = ['num_orders', 'meal_id', 'center_id', 'week', 'city_code', 'category', 'cuisine', 'center_type']
feature_cols = [col for col in df.columns if col not in exclude_cols]

X_train = train_data[feature_cols].fillna(0)
y_train = train_data['num_orders']
X_test = test_data[feature_cols].fillna(0)
y_test = test_data['num_orders']

print('Train shape:', X_train.shape)
print('Test shape:', X_test.shape)
```

```python
linear_model = LinearRegression()
linear_model.fit(X_train, y_train)
y_pred_linear = linear_model.predict(X_test)
linear_mae = mean_absolute_error(y_test, y_pred_linear)
linear_rmse = mean_squared_error(y_test, y_pred_linear) ** 0.5

poly = PolynomialFeatures(degree=2, include_bias=False)
X_train_poly = poly.fit_transform(X_train)
X_test_poly = poly.transform(X_test)
poly_model = LinearRegression()
poly_model.fit(X_train_poly, y_train)
y_pred_poly = poly_model.predict(X_test_poly)
poly_mae = mean_absolute_error(y_test, y_pred_poly)
poly_rmse = mean_squared_error(y_test, y_pred_poly) ** 0.5

results = pd.DataFrame({
    'Model': ['Linear Regression', 'Polynomial Regression'],
    'MAE': [linear_mae, poly_mae],
    'RMSE': [linear_rmse, poly_rmse]
})
results
```

```python
if poly_mae < linear_mae:
    chosen_model_name = 'Polynomial Regression'
    y_pred_final = y_pred_poly
else:
    chosen_model_name = 'Linear Regression'
    y_pred_final = y_pred_linear

print('Chosen model:', chosen_model_name)

plt.figure(figsize=(12, 5))
plt.plot(test_data['week'].values, y_test.values, label='Actual Orders', marker='o')
plt.plot(test_data['week'].values, y_pred_final, label='Predicted Orders', marker='s')
plt.title('Actual vs Predicted Orders on Test Period')
plt.xlabel('Week')
plt.ylabel('Number of Orders')
plt.legend()
plt.show()
```

## Reflection

This model is useful because it directly uses business-related features such as promotions and prices. Its main strength is interpretability, while its limitation is that it assumes future external variables are known in advance.
